## 1. Carga de Datos y Exploración

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix

# TODO: Carga el dataset WBCD desde UCI
# df_wbcd = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data", header=None)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data"
df_wbcd = pd.read_csv(url, header=None)
# Agrega nombres de columnas si es necesario
column_names = ['id', 'diagnosis'] + [f'feature_{i}' for i in range(1, 31)]
df_wbcd.columns = column_names

print("Dataset cargado exitosamente. Dimensiones:", df_wbcd.shape)
```

## 2. Preprocesamiento

```python
# TODO: Convierte la columna de diagnosis (M/B) a binario (1/0)
y = df_wbcd['diagnosis'].map({'M': 1, 'B': 0})
# TODO: Separa en X (features) e y (target)
X = df_wbcd.drop(columns=['id', 'diagnosis'])
# TODO: Haz el train_test_split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print(f"Datos de entrenamiento: {X_train.shape[0]}, Datos de prueba: {X_test.shape[0]}")
```

## 3. Escalamiento (Solo para k-NN)

```python
# TODO: Aplica StandardScaler a X_train y X_test.
# Recuerda: .fit_transform() en train, y solo .transform() en test.
scaler = StandardScaler()
X_train_esc = scaler.fit_transform(X_train)
X_test_esc = scaler.transform(X_test)

print("Escalamiento completado con éxito.")
```

## 4. Modelado y Matrices de Confusión

```python
# TODO: Entrena un KNeighborsClassifier(n_neighbors=5)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_esc, y_train)
pred_knn = knn.predict(X_test_esc)
# TODO: Entrena un DecisionTreeClassifier(max_depth=4)
arbol = DecisionTreeClassifier(max_depth=4, random_state=42)
arbol.fit(X_train, y_train)
pred_arbol = arbol.predict(X_test)
# TODO: Imprime las matrices de confusión para ambos
print("--- Matriz de Confusión: k-NN ---")
print(confusion_matrix(y_test, pred_knn))

print("\n--- Matriz de Confusión: Árbol de Decisión ---")
print(confusion_matrix(y_test, pred_arbol))
```

1. **Error Tipo II (Falso Negativo):** Ocurre cuando el modelo predice que un tumor es benigno (0) cuando en realidad es maligno (1). En una clínica oncológica, este error es crítico porque significa que un paciente enfermo no recibirá su diagnóstico ni tratamiento a tiempo, lo que puede permitir que el cáncer avance y ponga en riesgo su vida.

2. **Inmunidad del Árbol de Decisión al escalamiento:** Los árboles de decisión toman decisiones basadas en reglas lógicas de umbrales ("mayor que" o "menor que") ordenando de forma individual los valores de cada característica. Por lo tanto, no les afecta la magnitud absoluta ni las diferentes escalas numéricas de las variables, ya que el orden relativo de los datos se mantiene intacto.
