# Guía de práctica: predicción del costo de un trabajo de impresión

## Escenario

Un taller de impresión y un estudio de diseño gráfico cotizan carteles, flyers, lookbooks, banners y empaques. Cada fila es un **trabajo cerrado**. El objetivo es estimar el **costo final en dólares** (`total_cost_usd`) a partir de variables conocidas al cotizar: tiraje, gramaje del papel, páginas a color, área de recorte, cobertura de tinta, barniz y hora de registro de la cotización.

Se elige **costo de producción** y no “calidad percibida del diseño” porque el costo es una magnitud continua con drivers aditivos (más ejemplares, más área, más acabados). Eso hace que la **regresión lineal** sea un modelo razonable. Una calificación 1–5 o “aceptado / rechazado” exigiría otro tipo de modelo.

## Diccionario de columnas

| Columna | Tipo/unidad | Significado | Papel inicial |
|---|---|---|---|
| `job_id` | Texto | Identificador anónimo del trabajo | Identificador; no entra al modelo |
| `product_type` | Categoría | Poster, Flyer, Lookbook, Banner o Empaque | Contexto; requeriría codificación para modelar |
| `quantity` | Ejemplares | Tiraje impreso | Predictor numérico potencial |
| `paper_gsm` | g/m² | Gramaje del soporte | Predictor numérico potencial |
| `color_pages` | Entero | Páginas o caras a color | Predictor numérico potencial |
| `trim_area_sqin` | pulgadas² | Área de recorte (`ancho × alto`) | Predictor numérico potencial |
| `ink_coverage_pct` | 0–100 % | Cobertura media de tinta | Predictor numérico potencial |
| `coating` | 0/1 | 0 = sin barniz; 1 = barniz UV mate/brillo | Predictor numérico potencial |
| `quote_hour` | 8–18 h | Hora en que se registró la cotización | Predictor numérico potencial |
| `total_cost_usd` | USD | Costo final facturado | **Variable objetivo (`y`)** |

El conjunto es sintético y reproducible. Las asociaciones encontradas no demuestran causalidad. En un proyecto real, las variables deben estar disponibles **antes** de fijar el precio; no use costos posteriores ni descuentos negociados después de imprimir.

## 7. Explicar los resultados mediante métricas

### Funciones y fórmulas

- `mean_absolute_error(y_real, y_pred)`: MAE, error absoluto promedio en dólares.
- `mean_squared_error(...)`: MSE, promedio del error al cuadrado.
- `np.sqrt(MSE)`: RMSE, vuelve a dólares y penaliza más los errores grandes.
- `r2_score(y_real, y_pred)`: R², proporción de variabilidad explicada respecto de predecir siempre la media.

$$MAE=\frac{1}{n}\sum|y_i-\hat y_i|$$

$$RMSE=\sqrt{\frac{1}{n}\sum(y_i-\hat y_i)^2}$$

$$R^2=1-\frac{\sum(y_i-\hat y_i)^2}{\sum(y_i-\bar y)^2}$$

Menor MAE/RMSE es mejor; mayor R² es mejor. Compare entrenamiento y prueba: una brecha grande puede indicar sobreajuste.
