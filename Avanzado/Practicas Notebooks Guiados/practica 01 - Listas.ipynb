### Exercise 1.4 — List comprehension

Given:
```python
hours_played = [12, 45, 8, 120, 30]
```
- Build a list `double_hours` with each value multiplied by 2.
- Build a list `long_games` with only the values greater than 20.

**Documentation & how to solve**

- **List comprehension:** `[expression for item in list]` builds a new list. Add `if condition` to filter: `[x for x in list if x > 20]`.
- Double each: `[h * 2 for h in hours_played]`. Only > 20: `[h for h in hours_played if h > 20]`.
- **Python docs:** [List Comprehensions](https://docs.python.org/3/tutorial/datastructures.html#list-comprehensions)

**Sample solution:**

````python
# Sample solution for 1.4
hours_played = [12, 45, 8, 120, 30]
double_hours = [h * 2 for h in hours_played]
long_games = [h for h in hours_played if h > 20]
print("Double hours:", double_hours)
print("Long games (hours > 20):", long_games)
````

### Exercise 3.1 — Filter by label

Using your `samples` and `labels`:

**Documentation & how to solve**

### Exercise 3.2 — Simple train/test split

Split your data into "train" and "test" using list slicing (no randomness).
- Use the first 75% of `samples` and `labels` for training.
- Use the last 25% for test.

Hint: `n = len(samples)`, then use `n * 3 // 4` for the split index.

````python
# n = len(samples)
# split = n * 3 // 4
# X_train = samples[:split]
# y_train = labels[:split]
# X_test = samples[split:]
# y_test = labels[split:]
# print("Train:", len(X_train), "Test:", len(X_test))
````

**Documentation & how to solve**

- **Slicing for split:** `list[:split]` = first `split` elements, `list[split:]` = rest. Use same `split` for both `samples` and `labels`.
- **Integer division:** `n * 3 // 4` gives 75% of length (e.g. 4 → 3, 8 → 6). So train = first 75%, test = last 25%.

**Sample solution:**

````python
# Sample solution for 3.2
samples = [[10, 5, 2], [50, 20, 15], [3, 1, 0], [80, 45, 30]]
labels = ["abandoned", "completed", "abandoned", "completed"]
n = len(samples)
split = n * 3 // 4
X_train = samples[:split]
y_train = labels[:split]
X_test = samples[split:]
y_test = labels[split:]
print("Train:", len(X_train), "samples, Test:", len(X_test), "samples")
print("X_train:", X_train, "y_train:", y_train)
````

### Exercise 3.3 — Extract one feature from all samples

From `samples` (list of lists), build a list of just the first feature (e.g. `hours_played`) for every sample.

Example: if `samples = [[10,5,2], [50,20,15], ...]` then `hours = [10, 50, ...]`.

Use a list comprehension. Then compute `min`, `max`, and `sum` of that list.

````python
# hours = [row[0] for row in samples]
# print("Hours - min:", min(hours), "max:", max(hours), "sum:", sum(hours))
````

**Documentation & how to solve**

- **Extract one “column”:** for list of lists, first feature of each row = `row[0]`. So: `[row[0] for row in samples]`.
- **Built-ins:** `min(list)`, `max(list)`, `sum(list)` work on lists of numbers.

**Sample solution:**

````python
# Sample solution for 3.3
samples = [[10, 5, 2], [50, 20, 15], [3, 1, 0], [80, 45, 30]]
hours = [row[0] for row in samples]
print("Hours - min:", min(hours), "max:", max(hours), "sum:", sum(hours))
````

**Documentation & how to solve**

- **Count bought:** `bought.count(1)` or `sum(bought)` if bought = 0/1.
- **Scores of bought games:** use comprehension with index or zip: `[game_features[i][1] for i in range(len(bought)) if bought[i] == 1]`, then `sum(...)/len(...)` for average (guard against empty list).

**Sample solution:**

````python
# Sample solution for Part 4
game_features = [
    [20, 85, 0], [15, 72, 1], [60, 90, 0], [10, 68, 1], [40, 78, 0], [5, 88, 1]
]
bought = [0, 1, 1, 0, 1, 0]
bought_count = sum(bought)
scores_of_bought = [game_features[i][1] for i in range(len(bought)) if bought[i] == 1]
avg_score = sum(scores_of_bought) / len(scores_of_bought) if scores_of_bought else 0
print("Bought:", bought_count, "Average score (bought):", round(avg_score, 1))
````

## Quick reference: lists → ML

| Concept        | List example                          | In ML (e.g. scikit-learn) |
|----------------|---------------------------------------|----------------------------|
| Features      | `[[1,2], [3,4], [5,6]]`               | `X` (2D array)             |
| Labels        | `["A", "B", "A"]` or `[0, 1, 0]`   | `y`                        |
| One sample    | `[1, 2]`                              | One row of `X`            |
| Train/test    | `X[:split]`, `X[split:]`               | `train_test_split(X, y)`   |
| Filter by col | `[row[0] for row in X]`                | `X[:, 0]` in NumPy        |

Once you're comfortable with these list operations, the same ideas carry over to NumPy arrays and pandas DataFrames used in real ML.
