**Sample (videogame context)** — Given a tuple of game names, slice first 2, last 2, and every other.

### 8. Tuples as dictionary keys

**Documentation**
- Dict keys must be **hashable** (immutable). Tuples are hashable; lists are not. So `(a, b)` can be a key; `[a, b]` cannot.
- Useful for composite keys: (department, room), (test_name, date).

**Sample (videogame context)** — A dict with (genre, platform) as key and game count as value.

### 9. Zip and tuples

**Documentation**
- `zip(a, b)` pairs elements; each pair is a **tuple**. So `for x, y in zip(a, b):` unpacks tuples.
- [Python docs: zip()](https://docs.python.org/3/library/functions.html#zip)

**Sample (videogame context)** — Zip game names and scores, then loop and print each tuple.

```python
# Sample: zip returns tuples
games = ["Zelda", "Minecraft", "Elden Ring"]
scores = [95, 88, 96]
for pair in zip(games, scores):
    print(pair, type(pair))
for game, score in zip(games, scores):
    print(f"{game}: {score}")
```

## Quick reference: Tuples

| Concept | Syntax / use |
|--------|---------------|
| Create | `t = (a, b, c)` or `t = a, b, c` |
| Single element | `(x,)` (comma required) |
| Index / slice | `t[0]`, `t[-1]`, `t[:2]` |
| Unpack | `a, b = t` |
| Return multiple | `return a, b` → caller gets `(a, b)` |
| From list | `tuple(my_list)` |
| Dict key | `d[(x, y)] = value` (tuples are hashable) |
| zip | `for a, b in zip(list1, list2):` (pairs are tuples) |
| ML shape | `n_samples, n_features = data.shape` (NumPy/pandas) |
