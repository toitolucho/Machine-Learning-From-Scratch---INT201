### 3. get() and default value

**Documentation**
- `d.get(key)` returns `None` if key is missing. `d.get(key, default)` returns `default` instead.
- Avoids `KeyError` when the key might not exist. [Python docs: get](https://docs.python.org/3/library/stdtypes.html#dict.get)

**Sample (videogame context)** — Use `.get()` to safely look up a game score; use a default of `0` if the game is not in the dict.

```python
# Sample: get with default
game_scores = {"Elden Ring": 96, "Minecraft": 88}
score1 = game_scores.get("Elden Ring", 0)
score2 = game_scores.get("Unknown Game", 0)
print("Elden Ring:", score1, "| Unknown:", score2)
```

## Part 2: Iteration — keys, values, items

### 4. keys(), values(), items()

**Documentation**
- `d.keys()` — iterable of keys. `d.values()` — iterable of values. `d.items()` — iterable of (key, value) pairs.
- Often used in loops: `for k in d:`, `for v in d.values():`, `for k, v in d.items():`.

**Sample (videogame context)** — Print all keys, then all values, then each (game, score) using `.items()`.

```python
# Sample: keys, values, items
game_scores = {"Elden Ring": 96, "Minecraft": 88, "Celeste": 92}
print("Keys:", list(game_scores.keys()))
print("Values:", list(game_scores.values()))
for game, score in game_scores.items():
    print(f"  {game}: {score}")
```

### 6. The `in` operator

**Documentation**
- `key in d` is `True` if the key exists. `key not in d` for the opposite. Checks keys only, not values.

## Part 3: Building and transforming dicts

### 7. Dict from zip (two lists -> dict)

**Documentation**
- `dict(zip(keys_list, values_list))` builds a dict by pairing first with first, second with second, etc.
- Useful when you have parallel lists (e.g. IDs and names).

**Sample (videogame context)** — Build a dict mapping game abbreviations to full names using two lists and `zip`.

```python
# Sample: dict from zip
abbrevs = ["ER", "MC", "Cel"]
names = ["Elden Ring", "Minecraft", "Celeste"]
abbrev_to_name = dict(zip(abbrevs, names))
print(abbrev_to_name)
```

### 8. Dict comprehension

**Documentation**
- `{key_expr: value_expr for item in iterable}` or `for k, v in d.items()` with optional `if`. Builds a new dict.
- [Python docs: Dict comprehensions](https://docs.python.org/3/tutorial/datastructures.html#dictionaries)

**Sample (videogame context)** — From a score dict, build a new dict that doubles each score (game -> score*2).

```python
# Sample: dict comprehension
game_scores = {"Elden Ring": 96, "Minecraft": 88, "Celeste": 92}
doubled = {game: score * 2 for game, score in game_scores.items()}
print(doubled)
```

## Part 4: Update/merge and practical use

### 10. Update / merge dicts

**Documentation**
- `d1.update(d2)` adds all key-value pairs from `d2` into `d1`. If a key exists, its value is overwritten. For a new dict without changing originals: `{**d1, **d2}` (Python 3.5+).

**Sample (videogame context)** — Merge a second score dict into the first with `.update()`.

```python
# Sample: update (merge)
game_scores = {"Elden Ring": 96, "Minecraft": 88}
more_scores = {"Hades": 90, "Minecraft": 89}
game_scores.update(more_scores)
print(game_scores)
```

## Quick reference: Dictionaries

| Concept | Syntax / use |
|--------|---------------|
| Create | `d = {"a": 1, "b": 2}` |
| Access | `d["a"]` |
| Safe access | `d.get("a", default)` |
| Add/update | `d["c"] = 3` |
| Keys/values/items | `d.keys()`, `d.values()`, `d.items()` |
| Check key | `"a" in d` |
| From lists | `dict(zip(keys, values))` |
| Comprehension | `{k: v*2 for k, v in d.items()}` |
| Nested | `d["key"]["inner"]` |
| Merge | `d1.update(d2)` or `{**d1, **d2}` |
| Copy | `d.copy()` or `dict(d)` |
