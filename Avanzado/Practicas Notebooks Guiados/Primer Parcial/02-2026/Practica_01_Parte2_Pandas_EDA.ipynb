## Exercise 2.2 -- Distribution of Numerical Features and Outlier Detection

**Reference.** Visualizing feature distributions reveals central tendencies, spread, skewness, and potential outliers:

## Exercise 2.4 -- Correlation Matrix and Heatmap Visualization

**Reference.** Pearson's correlation coefficient $r \in [-1, 1]$ measures the strength and direction of the linear relationship between two continuous variables:
$$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$
- $r \approx +1.0$: Strong positive linear correlation (as $x$ increases, $y$ increases linearly).
- $r \approx -1.0$: Strong negative linear correlation (as $x$ increases, $y$ decreases linearly).
- $r \approx 0.0$: Little or no linear relationship.
- `df.corr()`: Computes pairwise correlations between all numeric columns.
- `sns.heatmap(corr, annot=True, cmap="coolwarm")`: Color-codes the matrix for quick visual scanning.

**Example.** The cell below computes and displays a correlation matrix on a toy dataset.

```python
# Example -- do not modify
demo_corr = pd.DataFrame({"x": [1, 2, 3, 4], "y": [2, 4, 6, 8], "z": [10, 8, 5, 1]}).corr()
print("Correlation matrix:\n", demo_corr.round(2))
```

## Exercise 2.5 -- Categorical Variable Encoding

**Reference.** Machine learning algorithms require numerical inputs. Categorical variables must be converted to numeric representations:
- **One-Hot Encoding (Nominal):** Converts categorical columns with $k$ categories into $k$ binary indicators ($0$ or $1$). In regression models, `drop_first=True` drops the first column to create $k-1$ indicators, avoiding perfect multicollinearity (the **dummy variable trap**).
- **Ordinal Encoding (Ordered):** When categories follow a natural progression or complexity order, a dictionary mapping can assign sequential integers: `df["col"].map({"low": 0, "med": 1, "high": 2})`.

**Example.** The cell below illustrates dummy encoding with `drop_first=True` and manual ordinal mapping.

```python
# Example -- do not modify
demo_df = pd.DataFrame({"role": ["support", "protagonist", "antagonist"], "level": ["low", "medium", "high"]})
print("Dummies with drop_first:\n", pd.get_dummies(demo_df[["role"]], drop_first=True))
level_map = {"low": 0, "medium": 1, "high": 2}
demo_df["level_num"] = demo_df["level"].map(level_map)
print("\nMapped ordinal:\n", demo_df[["level", "level_num"]])
```
