## Ejercicio 2.4 -- Matriz de Correlación y Mapa de Calor

**Referencia.** El coeficiente de correlación de Pearson $r \in [-1, 1]$ cuantifica la fuerza y dirección de la relación lineal entre dos variables continuas:
$$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$
- $r \approx +1.0$: Relación lineal positiva muy fuerte (a mayor $x$, mayor $y$).
- $r \approx -1.0$: Relación lineal negativa muy fuerte (a mayor $x$, menor $y$).
- $r \approx 0.0$: Inexistencia de relación lineal.
- `df.corr()`: Calcula correlaciones cruzadas entre todas las columnas numéricas.
- `sns.heatmap(corr, annot=True, cmap="coolwarm")`: Visualiza la matriz con escala de color.

**Ejemplo.** La siguiente celda calcula la matriz de correlación sobre datos sintéticos.

```python
# Ejemplo -- no modificar
demo_corr = pd.DataFrame({"x": [1, 2, 3, 4], "y": [2, 4, 6, 8], "z": [10, 8, 5, 1]}).corr()
print("Matriz de correlacion:\n", demo_corr.round(2))
```

```python
# Solucion 2.4
df_num = df.select_dtypes(include="number")
corr_matrix = df_num.corr()

# Mapa de calor
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Matriz de Correlacion — Dataset Neon Skies", fontsize=14)
plt.tight_layout()
plt.show()

# Correlaciones con el target ordenadas descendentemente
corr_target = corr_matrix["production_hours"].drop("production_hours").sort_values(ascending=False)
print("Correlacion de Pearson con production_hours:")
print(corr_target.round(4))
```

## Ejercicio 2.5 -- Codificación de Variables Categóricas

**Referencia.** Los modelos de Machine Learning requieren entradas estrictamente numéricas:
- **One-Hot Encoding (Nominal):** Convierte una columna categórica con $k$ categorías en variables binarias (0 o 1). En modelos de regresión, `drop_first=True` elimina la primera categoría para evitar multicolinealidad perfecta (la **trampa de la variable dummy**).
- **Mapeo Ordinal (Jerárquico):** Cuando existe un orden inherente de complejidad artística (flat < cel_shaded < painted), se asignan números enteros secuenciales con `.map({"categoria": valor})`.

**Ejemplo.** La siguiente celda ilustra `drop_first=True` y mapeo manual.

```python
# Ejemplo -- no modificar
demo_df = pd.DataFrame({"rol": ["support", "protagonist", "antagonist"], "nivel": ["bajo", "medio", "alto"]})
print("Dummies con drop_first:\n", pd.get_dummies(demo_df[["rol"]], drop_first=True))
nivel_map = {"bajo": 0, "medio": 1, "alto": 2}
demo_df["nivel_num"] = demo_df["nivel"].map(nivel_map)
print("\nMapeo ordinal:\n", demo_df[["nivel", "nivel_num"]])
```

**Tarea.** Completa la celda del estudiante:
1. Aplica one-hot encoding a `role` con `pd.get_dummies(df, columns=["role"], drop_first=True)` e imprime las columnas resultantes.
2. Define el diccionario de mapeo ordinal para estilo de animación: `style_map = {"flat": 0, "cel_shaded": 1, "painted": 2}`.
3. Mapea la columna `style` a una nueva columna `df["style_code"]` usando `.map(style_map)`.
4. Muestra las combinaciones únicas de `style` y `style_code` ordenadas por `style_code`.

```python
# Solucion 2.5
df_encoded = pd.get_dummies(df, columns=["role"], drop_first=True)
print("Columnas tras get_dummies (drop_first=True):")
print(df_encoded.columns.tolist())

# Mapeo ordinal para style
style_map = {
    "flat": 0,
    "cel_shaded": 1,
    "painted": 2
}
df["style_code"] = df["style"].map(style_map)

print("\nVerificacion de mapeo de style_code:")
print(df[["style", "style_code"]].drop_duplicates().sort_values(by="style_code"))
```

**Pregunta Lógica 2.5 -- Justificación y Respuesta:**

**a) ¿Por qué es necesario usar `drop_first=True` al crear variables dummy para regresión lineal?**
**Respuesta:** Evita la trampa de la variable dummy (multicolinealidad perfecta). Si se conservan las $k$ categorías junto con el término de intercepción (bias), la suma de las columnas dummy suma 1 exactamente, provocando que la matriz $(X^T X)$ sea singular y no invertible en la Ecuación Normal.

**b) Si la columna `role` tiene 3 categorías (`antagonist`, `protagonist`, `support`), ¿cuántas columnas dummy se generan con `drop_first=True` y cuál categoría actúa como referencia base?**
**Respuesta:** Se generan exactamente 2 columnas dummy: `role_protagonist` y `role_support`. La categoría `antagonist` (primera alfabéticamente) se omite y actúa como la categoría base de referencia (cuando ambas dummies son 0).

**c) En el mapeo de `style_code`, ¿qué valor numérico entero se asigna al personaje `Lyra`?**
**Respuesta:** `1` (ya que el estilo de Lyra es `cel_shaded`, asignado al código 1).

## Ejercicio 2.6 -- Selección de Características y Construcción del Dataset Final

**Referencia.** La selección de características reduce sobreajuste y simplifica el modelo eliminando atributos sin poder predictivo:
- **Umbral de Correlación:** Retener aquellas características con $|r| \ge \text{umbral}$ frente a la variable objetivo.
- **Eliminación de Redundancias:** Si dos características aportan información equivalente (`num_colors` y `num_frames_total`), se descarta la redundante para evitar multicolinealidad.
- **Exportación:** El dataset final se guarda en disco con `df.to_csv("nombre.csv", index=False)`.

**Ejemplo.** La siguiente celda ilustra el filtrado por umbral de correlación.

```python
# Ejemplo -- no modificar
corrs = pd.Series({"frames": 0.98, "tiempo": 0.95, "ruido": 0.12})
umbral = 0.60
seleccionadas = corrs[corrs.abs() >= umbral].index.tolist()
print("Caracteristicas que superan el umbral 0.60:", seleccionadas)
```

**Tarea.** Completa la celda del estudiante:
1. Define el umbral de correlación: `threshold = 0.60`.
2. Filtra las características de `corr_target` cuyo valor absoluto $|r| \ge \text{threshold}$ (excluyendo `production_hours`).
3. Descarta `num_colors` para evitar redundancia con `num_frames_total`.
4. Construye `df_features` con `name`, las características predictoras seleccionadas y `production_hours`.
5. Muestra `df_features.head(5)` y `df_features.describe().round(2)`.
6. Exporta `df_features` al archivo `characters_features.csv` con `index=False`.

```python
# Solucion 2.6
threshold = 0.60

# Filtrar caracteristicas por umbral de correlacion
selected_features = [col for col in corr_target.index if abs(corr_target[col]) >= threshold]
# Excluir num_colors para evitar redundancia con num_frames_total
if "num_colors" in selected_features:
    selected_features.remove("num_colors")

print("Caracteristicas predictoras seleccionadas (|r| >= 0.60, no redundantes):")
print(selected_features)

# Construir DataFrame final de caracteristicas
feature_dataset_cols = ["name"] + selected_features + ["production_hours"]
df_features = df[feature_dataset_cols].copy()

print("\nVista previa del dataset final (primeras 5 filas):")
print(df_features.head(5))

print("\nResumen estadistico del dataset final:")
print(df_features.describe().round(2))

# Exportar a CSV
output_filename = "characters_features.csv"
df_features.to_csv(output_filename, index=False)
print(f"\nExportado exitosamente: {output_filename} con dimensiones {df_features.shape}")
```
