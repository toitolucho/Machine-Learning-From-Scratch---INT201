# Clase Práctica 4: Regresión y Expansión Polinómica

Hoy implementaremos la regresión lineal para predecir variables continuas (como los FPS o tiempo de renderizado de un videojuego). Usaremos transformaciones polinómicas y veremos cómo esto nos lleva al temido *Overfitting*.

## 1. El Cordel Elástico (Regresión Lineal Simple)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Simulamos datos: Complejidad de la Escena 3D (X) vs Tiempo de Render (y)
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

lin_reg = LinearRegression()
lin_reg.fit(X, y)

print(f"Intercepto: {lin_reg.intercept_[0]:.2f}")
print(f"Coeficiente (Pendiente): {lin_reg.coef_[0][0]:.2f}")
```

## 2. La Montaña Rusa (Regresión Polinómica)

```python
from sklearn.preprocessing import PolynomialFeatures

# Creando datos curvos
X_curvo = 6 * np.random.rand(100, 1) - 3
y_curvo = 0.5 * X_curvo**2 + X_curvo + 2 + np.random.randn(100, 1)

# Expansión de características
poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(X_curvo)

lin_reg_poly = LinearRegression()
lin_reg_poly.fit(X_poly, y_curvo)

print("Modelo Polinómico Ajustado:")
print(f"Coeficientes: {lin_reg_poly.coef_}, Intercepto: {lin_reg_poly.intercept_}")
```

## 3. Diagnóstico de Multicolinealidad (VIF)
Si le pasamos características duplicadas a la computadora, el modelo se confunde.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

# Dataset de ejemplo
df = pd.DataFrame({
    'resolucion_ancho': [1920, 1280, 2560, 3840],
    'resolucion_alto': [1080, 720, 1440, 2160], # Totalmente lineal a ancho (16:9)
    'poligonos': [1000, 500, 2000, 3000]
})

vif_data = pd.DataFrame()
vif_data["Atributo"] = df.columns
vif_data["VIF"] = [variance_inflation_factor(df.values, i) for i in range(len(df.columns))]
print(vif_data)

# Un VIF > 10 indica un problema grave.
```
